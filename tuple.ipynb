```python
#Q1. Write a program to create a tuple with numbers and print the 4th element from the first and the 4th element from the last.
numbers=(10,20,30,40,50,60,70,80,90)
print("Tuple:" ,numbers)
print("4th element from the first:",numbers[3])
print("4th element from the Last:",numbers[-4])
```

```python
#Q2. Write a program to create a tuple with numbers and check whether an element is present in the tuple or not.
tup_1=(10,20,30,40,50,60,70,80,90)
element=int(input("Enter the element you want to find:"))
if element in tup_1:
    print("Element is present in the tuple")
else:
    print("Element is not present in the tuple")
```

```python
#Q3. Write a program to convert a list to a tuple.
list_1=[10,20,30,40,50,60,70,80,90]
tuple_1=tuple(list_1)
print("List:" ,list_1)
print("Tuple:" ,tuple_1)
```

```python
#Q4. Write a program to create a tuple with numbers and find the index of an element in the tuple.
tuple_2=(10,20,30,40,50,60,70,80,90)
element=int(input("Enter the element you want to find:"))
if element in tuple_2:
    print("Element is present in the tuple at index:",tuple_2.index(element))
else:
    print("Element is not present in the tuple")
```

```python
#Q5. Write a program to create a list of tuples and add 100 to the last element of each tuple in the list.
list_1=[(10,20,40),(40,50,60),(70,80,90)]
new_list=[]
for tup in list_1:
    new_tup= tup[:-1]+(100,)
    new_list.append(new_tup)
print("Original List:",list_1)
print("New List:",new_list)
```

```python
#Q6. Write a program to create a tuple with numbers and find the maximum and minimum values in the tuple.
number_tuple = (10,20,25,35,62,75,80,90,100)
max_value = max(number_tuple)
min_value = min(number_tuple)   
print("Maximum value in the tuple:", max_value)
print("Minimum value in the tuple:", min_value)
```
